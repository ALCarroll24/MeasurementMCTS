```python
def check_stop_distance(speed, dt):
    start_time = timeit.default_timer()
    state = [0,0, speed, 0, 0, 0]
    time = 0
    while state[2] > 0:
        state = env.car.update(dt, [-1,0], starting_state=state, model_dt=0.0)
        time += dt
    elapsed = timeit.default_timer() - start_time
    return time, state[0], elapsed, state[2]

print(check_stop_distance(8, 0.6))
print(check_stop_distance(8, 0.1))
print(check_stop_distance(8, 0.05))
print(check_stop_distance(8, 0.01))
print(check_stop_distance(8, 0.001))
print(check_stop_distance(8, 0.0001))
```
