```python
import numpy as np
```

```python
# Action space parameters
# Options for longitudinal acceleration (scaled from [-1, 1] to vehicle [-max_acc, max_acc])
long_acc_options = np.array([-1., -0.5, 0., 0.5, 1.]) 
# Options for steering acceleration (scaled from [-1, 1] to vehicle [-max_steering_alpha, max_steering_alpha])
steering_acc_options = np.array([-1., -0.25, 0., 0.25, 1.])
action_space = np.array(np.meshgrid(long_acc_options, steering_acc_options)).T.reshape(-1, 2) # Generate all combinations using the Cartesian product of the two action spaces
zero_action = np.array([0.0, 0.0]) # Move the zero action to the front of the list (this is the default action)
zero_action_index = np.where(np.all(action_space == zero_action, axis=1))[0][0]
action_space = np.concatenate((action_space[zero_action_index:], action_space[:zero_action_index]))
print("Number of actions: ", len(action_space))
print("Action space:")
print(action_space)
```
